```python
class Hospital():
    def __init__(self,P_Name,P_Age,Disease,Blood_Group):       # parametric constructor
        self.Patient=      P_Name  # attributes
        self.Age =         P_Age
        self.Disease =     Disease
        self.Blood_Group = Blood_Group
    def details(self):         # method 1
        print(f"Patient {self.Patient} whose age is {self.Age} has come for the treatment of {self.Disease}.")
    def Patient_Blood_Group(self):            # method 2]s
        print(f"Blood_Group of patient {self.Patient} is {self.Blood_Group}")
```

# INHERITANCE

```python
class employee():
    def __init__(self,ID,name,aadhar):    # constructor
        print("parent constructor has run")
        self.ID  = ID
        self.fullname = name
        self.id_proof = aadhar
    def welcome(self):
        print(f"welcome {self.fullname}")

class department(employee):     # child class (after putting the class() of child in the class in the bracket will consider as parent        
    def __init__(self,ID,name,aadhar,dept,salary):   # if there is a same method (ie.info) in both the class child class will ignore the parent class.
        print("child constructor has run")
        super().__init__(ID,name,aadhar)   # super function calls the parent's constructor (will read all these attributes from parent class)
        self.department = dept               # child class have full access to parent class but not vice-versa
        self.payment = salary                # you can't add extract obj/attribute  in parent class you have to add in child class always
    def info(self):
        print(f"the department of {self.fullname} is {self.department}")
        print(f"the diwali bonus of {self.fullname} is {(self.payment*0.05)}")
```

```python
p1 = employee(1001,"john doe",1234354)
```

```python
c1 = department(1002,"mary smith",1234568,"hr",500000)    # both the line run because of super function
```

```python
p1. welcome()
p1. info()       # parent class object has no access over child class
```

```python
c1. welcome()      # child class have full access on parent class
c1. info() 
```

# ABSTRACTION

```python
from abc import ABC,abstractmethod
```

```python
class company(ABC):     # after writting Abc in the bracket now it become abstract-parent class
    @ abstractmethod     # due to this the child class has to access according to the parent class
    def info(self):       # abstract method 1
        print("information")
    def details(self):     # method 2
        print("details")

class dept_hr(company):    # child class
    def kpi(self):
        print("kpi")
    def info(self):
        print("hello hr")

class dept_it(company):    # child class
    def kpi(self):
        print("kpi")
    def info(self):
        print("hello it")

it_001 = dept_hr()      # object creation
it_001.kpi()         # object cannot use the class unless "info method" is implemented in child class.
it_002= dept_it()
it_002.info()
```
