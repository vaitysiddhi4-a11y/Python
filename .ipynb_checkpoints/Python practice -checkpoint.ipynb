```python
x = 45.2
print(x)
print(type(x))

x = int(x)     # Type Casting
print(x)
print(type(x))
```

```python
x = "45"
print(x)
print(type(x))

x = int(x)     # Type Casting/ only pure no.
print(x)
print(type(x))
```

```python
x = "45.8"
print(x)
print(type(x))

x = float(x)     # Type Casting/ only float string can change to float
print(x)
print(type(x))

x = int(x)     # Type Casting/ float to int
print(x)
print(type(x))

x = bool(x)     # Type Casting/ int to bool (anything accept than zero is true)
print(x)
print(type(x))

x = complex(x)     # Type Casting/bool to complex (int with imagnary string)
print(x)
print(type(x))
```

# CONDITIONAL STATEMENTS

# if statements

```python
x=20
print(x>0)


if x>0:       #colen is complusary
    print("Number is Positive")       #from , if till the condition written ,it is a code box   
```

```python
x=0
print(x>0)


if x>0:       #colen is complusary
    print("Number is Positive")      # if the condition is not true the curser will not entry into the code box

elif x==0:
    print("Number is Zero")

else:
    print ("Number is Negative")      # if and elif ,condition get false than only ,else is the by default to be taken (no option)

print("Program ends")   # after backspace the code box is finished
```

```python
x=700
# print(x==0)


if x>1000:       
    print("Number is Greater")      

elif x>500:
    print("Number lies between 500 and 1000")

else:
    print ("Number is less than 500")     

print("Program ends")
```
