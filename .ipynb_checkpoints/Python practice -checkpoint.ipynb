```python
x=20
print(x>0)


if x>0:       #colen is complusary
    print("Number is Positive")       #from , if till the condition written ,it is a code box   
```

```python
x=0
print(x>0)


if x>0:       #colen is complusary
    print("Number is Positive")      # if the condition is not true the curser will not entry into the code box

elif x==0:
    print("Number is Zero")

else:
    print ("Number is Negative")      # if and elif ,condition get false than only ,else is the by default to be taken (no option)

print("Program ends")   # after backspace the code box is finished
```

# PALINDROME

```python
w.a.p to print whether user entered number is a palindrome
```

```python
x=int(input("enter the number to check if it is palidrome: "))
x=str(x)
print(len(x))
list1=[]
for i in x:
    list1.append(i)
print(list1)

flag= False
for i in range(len(list1)):
    if list1[i]==list1[-1-i]:
        flag=False
    else:
        flag=True
        break
if flag==False:
    print("number is palindorme")
else:
    print("number is not palindorme")
    
        
    
```
