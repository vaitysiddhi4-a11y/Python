# PALINDROME

```python
w.a.p to print whether user entered number is a palindrome
```

```python
x=int(input("enter the number to check if it is palidrome: "))
x=str(x)
print(len(x))
list1=[]
for i in x:
    list1.append(i)
print(list1)

flag= False
for i in range(len(list1)):
    if list1[i]==list1[-1-i]:
        flag=False
    else:
        flag=True
        break
if flag==False:
    print("number is palindorme")         # palindorme means mirror effect such 343 from front and from back same effect
else:
    print("number is not palindorme")
    
        
    
```

# Recursive Function

```python
def fact(n):
    if n==1:
        return 1
    else:
        product = n * fact(n-1)
        return product
num = int(input("Enter number to find its factorial: "))
print(f"The factorial of {num} is {fact(num)}" )
print("The factorial of",num,"is",fact(num))
```
